```python
import os
import nibabel as nib
import numpy as np

def center_crop_3d(image: np.ndarray, crop_size=(256, 256, 256)) -> np.ndarray:
    """Crop the center of a 3D volume."""
    d, h, w = image.shape
    cd, ch, cw = crop_size
    sd = max((d - cd) // 2, 0)
    sh = max((h - ch) // 2, 0)
    sw = max((w - cw) // 2, 0)
    return image[sd:sd + cd, sh:sh + ch, sw:sw + cw]

def crop_and_save_nifti(input_dir: str, output_dir: str, crop_size=(256, 256, 256)):
    """Center crop all .nii or .nii.gz files in a directory and save to another directory."""
    os.makedirs(output_dir, exist_ok=True)
    for fname in os.listdir(input_dir):
        if fname.endswith('.nii') or fname.endswith('.nii.gz'):
            input_path = os.path.join(input_dir, fname)
            output_path = os.path.join(output_dir, fname)

            img = nib.load(input_path)
            data = img.get_fdata()
            affine = img.affine
            cropped = center_crop_3d(data, crop_size)

            cropped_img = nib.Nifti1Image(cropped, affine)
            nib.save(cropped_img, output_path)
            print(f"Cropped and saved: {output_path}")
    print("croped")     

# Example usage
if __name__ == "__main__":
    input_dir = "../dataset/TRAINCTAlignedToCBCT"
    output_dir = "../croped_dataset/TRAINCTAlignedToCBCT"
    crop_size = (256, 256, 256)  # You can adjust as needed

    crop_and_save_nifti(input_dir, output_dir, crop_size)
```
